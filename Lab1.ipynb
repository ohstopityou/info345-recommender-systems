```python
# Load Libraries
import pandas as pd
```

- read the users data

```python
from pathlib import Path
from urllib.request import Request, urlopen
from zipfile import ZipFile
import shutil
import ssl

# Download MovieLens 100K once per fresh Colab runtime.
data_dir = Path("/content/ml-100k")
if not data_dir.exists():
    archive = Path("/content/ml-100k.zip")
    request = Request(
        "https://files.grouplens.org/datasets/movielens/ml-100k.zip",
        headers={"User-Agent": "Mozilla/5.0"},
    )

    # GroupLens currently serves an expired TLS certificate.
    ssl_context = ssl._create_unverified_context()
    with urlopen(request, context=ssl_context) as response:
        with archive.open("wb") as output_file:
            shutil.copyfileobj(response, output_file)

    with ZipFile(archive) as zip_file:
        zip_file.extractall("/content")

u_cols = ["user_id", "age", "sex", "occupation", "zip_code"]
users = pd.read_csv(
    data_dir / "u.user",
    sep="|",
    names=u_cols,
    encoding="latin-1",
)
```

```python
# Read the item metadata.
i_cols = [
    "movie id", "movie title", "release date", "video release date",
    "IMDb URL", "unknown", "Action", "Adventure", "Animation",
    "Children's", "Comedy", "Crime", "Documentary", "Drama", "Fantasy",
    "Film-Noir", "Horror", "Musical", "Mystery", "Romance", "Sci-Fi",
    "Thriller", "War", "Western",
]
items = pd.read_csv(
    data_dir / "u.item",
    sep="|",
    names=i_cols,
    encoding="latin-1",
)
```

```python
ratings_per_movie = ratings.groupby("movie_id").size()

dataset_summary = pd.Series({
    "Users": users["user_id"].nunique(),
    "Movies": items["movie id"].nunique(),
    "Ratings": len(ratings),
    "Mean rating": ratings["rating"].mean(),
    "Mean ratings per movie": ratings_per_movie.mean(),
})
dataset_summary
```
